```python
mapping = {"never": 0, 'current': 1, "former":3,"No Info":4, "ever": 5 ,"not currer":6 }
df['smoking_history'] = df['smoking_history'].map(mapping)
```

```python
mapping={"Male":1,"Female":0}
df["gender"] =df["gender"].map(mapping)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=["gender"])
```

```python
df["smoking_history"]=df["smoking_history"].fillna(4)
```

```python
df.isnull().sum()
```

```python
X=df.drop("diabetes",axis=1)
Y=df["diabetes"]
```

```python
X
```

```python
Y
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2, random_state=42)
```

```python
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, Y_train)
```

```python
accuracy = model.score(X_test, Y_test)
print(f"Accuracy after training: {accuracy * 100:.2f}%")
```
